````python
def remove_extra_spaces(text):
    text = ' '.join(text.split())                    # collapse runs of whitespace
    text = re.sub(r'\s+([.,!?;:])', r'\1', text)     # remove space before punctuation
    return text
print(remove_extra_spaces(sample_text))
# This one doesn't split every character by a period.
````

````python
def remove_numbers(text):
    return ''.join([char for char in text if not char.isdigit()])

remove_extra_spaces(remove_numbers(sample_text))
# Wrapped it in the remove spaces.
````

### Exercise 2(d) (3 points)

Remove the punctuation from `sample_text`. Hint: you the same approach presented in Exercise 1(b)

````python
import string

def remove_punctuation(text):
    return ''.join([char for char in text if char not in string.punctuation])

print(remove_extra_spaces(remove_punctuation(sample_text)))
````

### Exercise 2(e) (4 points)

Using `set()`, report the unique words in `sample_text`.

````python
print('Unique Words:')
uw = set(remove_extra_spaces(remove_punctuation(remove_numbers(sample_text))).split(' '))
for word in sorted(uw):
    print(f'• {word}')
````

Explain the difference between **tokenization** and **stemming/lemmatization** in the context of text preprocessing. Provide a short example of each.

Tokenization is breaking text into smaller units called tokens. 

Example: 

"My name is Ben DuPey" -> ["My", "name", "is", "Ben", "DuPey"]

Stemming/lemmatizatization reduces words to base forms so that similar words are treated similarly.

Example:

["Teach", "Teaching", "Teaches"] -> ["Teach", "Teach", "Teach"]
