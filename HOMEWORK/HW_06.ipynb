In this exercise, we will use the Levenshtein edit distance to build a simple fuzzy-matching (spell-correction) tool for the `variety` column of the `winemag-data-130k-v2.csv` data file. Use the data frame `wine_df` created in Exercise 1.

### Exercise 3(a) (3 points)

Without using any edit-distance library (e.g., `rapidfuzz` or `python-Levenshtein`), implement the following two functions using dynamic programming (Wagner–Fischer algorithm):

1. `levenshtein_distance(u, v)`: returns the Levenshtein edit distance between the strings `u` and `v` (case-insensitive).
2. `levenshtein_similarity(u, v)`: returns the normalized similarity $1 - \frac{LD(u,v)}{\max(|u|,|v|)}$.

Test your functions on the pairs (`Believe`, `beleive`), (`Believe`, `bargain`), and (`Believe`, `Elephant`).

````python
def levenshtein_distance(u, v):
    u, v = u.lower(), v.lower()
    m, n = len(u), len(v)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            cost = 0 if u[i - 1] == v[j - 1] else 1
            dp[i][j] = min(dp[i - 1][j] + 1, dp[i][j - 1] + 1, dp[i - 1][j - 1] + cost)
    return dp[m][n]

def levenshtein_similarity(u, v):
    return 1 - levenshtein_distance(u, v) / max(len(u), len(v))

for pair in [('Believe', 'beleive'), ('Believe', 'bargain'), ('Believe', 'Elephant')]:
    print(pair, levenshtein_distance(*pair), round(levenshtein_similarity(*pair), 4))
````

### Exercise 3(b) (3 points)

Extract the unique values of the `variety` column of `wine_df` (ignoring missing values). Then, write a function `top_matches(query, choices, n=3)` that returns the `n` closest varieties to `query` according to the Levenshtein distance (break ties using the order in which the varieties appear). Display the distance and the normalized similarity of each match.

Use your function on the misspelled queries: `Cabernet Sauvignion`, `Chardonay`, and `Pinot Noirr`.

````python
choices = wine_df['variety'].dropna().unique().tolist()

def top_matches(query, choices, n=3):
    scored = [(c, levenshtein_distance(query, c), levenshtein_similarity(query, c)) for c in choices]
    scored.sort(key=lambda t: t[1])
    return scored[:n]

for q in ['Cabernet Sauvignion', 'Chardonay', 'Pinot Noirr']:
    print(q)
    for c, d, s in top_matches(q, choices):
        print(f'  {c}: distance={d}, similarity={s:.4f}')
````

### Exercise 3(c) (4 points)

1. Using `levenshtein_similarity`, write a function `autocorrect(query, choices, threshold=0.75)` that returns the best match only if its similarity is greater than or equal to `threshold`, and returns `None` otherwise.
2. Test it on `Reisling`, `Chardonay`, and `Xyzzy`.
3. In a markdown cell, explain why the result for `Reisling` is only barely above the threshold, and what refinement of Levenshtein distance would score it as a closer match.

````python
def autocorrect(query, choices, threshold=0.75):
    best = max(choices, key=lambda c: levenshtein_similarity(query, c))
    sim = levenshtein_similarity(query, best)
    return best if sim >= threshold else None

for q in ['Reisling', 'Chardonay', 'Xyzzy']:
    print(q, '->', autocorrect(q, choices))
````
