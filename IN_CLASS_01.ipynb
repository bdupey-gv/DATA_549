# Exercise 2 (3 points)

Create a `pandas` data-frame called `df` with the `time_sentences` list:

```
time_sentences = ["Monday: The doctor's appointment is at 2:45pm.", 
                  "Tuesday: The dentist's appointment is at 11:30 am.",
                  "Wednesday: At 7:00pm, there is a basketball game!",
                  "Thursday: Be back home by 11:15 pm at the latest.",
                  "Friday: Take the train at 08:10 am, arrive at 09:00am."]
```

````python
time_sentences = {'Sentences' : ["Monday: The doctor's appointment is at 2:45pm.", 
                  "Tuesday: The dentist's appointment is at 11:30 am.",
                  "Wednesday: At 7:00pm, there is a basketball game!",
                  "Thursday: Be back home by 11:15 pm at the latest.",
                  "Friday: Take the train at 08:10 am, arrive at 09:00am."]
}
df = pd.DataFrame(time_sentences)
df.head()
````

# Exercise 3 (2 points)

Report the number of characters of each of the sentences in the `df` data-frame.

````python
df['Sentences'].str.len()
````

# Exercise 4 (2 points)

Report the number of tokens for each sentence in the `df` data-frame.

````python
df['Sentences'].str.split().str.len()
````

# Exercise 5 (2 points)

Report the sentences that contain the word `appointment` in the `df` data-frame.

````python
# I went ahead and did this part for fun :)
df[df['Sentences'].str.contains('appointment', case=False, na=False)]
````

# Exercise 6 (3 points)

Replace weekdays with their 3 letter abbreviations.

````python
df['Sentences'] = df['Sentences'].replace({
    'Monday': 'Mon',
    'Tuesday': 'Tues', 
    'Wednesday': 'Wed',
    'Thursday': 'Thur', 
    'Friday': 'Fri'
}, regex=True)

df.head()
````
