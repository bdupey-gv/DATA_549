# Working With Text

## Exercise 1 (2 points)

Create an object called `text` with the following sentence:

```
Ethics are built right into the ideals and objectives of the United Nations
```

````python
text = 'Ethics are built right into the ideals and objectives of the United Nations'
````

## Exercise 2 (2 points)

Report the length of `text`.

````python
len(text)
````

## Exercise 3 (2 points)

Split `text` based on a single space and store the words in an object called `text_1`.

````python
text_1 = text.split(' ')
text_1
````

## Exercise 4 (2 points)

Report the length of `text_1`.

````python
len(text_1)
````

## Exercise 5 (2 points)

Report the words with more than 3 characters in `text_1`.

````python
[word for word in text_1 if len(word) > 3]
````

## Exercise 6 (2 points)

Report the capitalized words from `text_1`.

````python
[word for word in text_1 if word[0].isupper()]
````

## Exercise 7 (2 points)

Report the words that end in `s` from `text_1`.

````python
[word for word in text_1 if word.endswith('s')]
````

## Exercise 8 (3 points)

Using `set()` report the number of unique words from the following text:

```
To be or not to be
```

````python
text_2 = "To be or not to be"
len(set(text_2.split(' ')))
````

## Exercise 9 (3 points)

Convert the words from part 8 to lowercase and report the unique words using `set()`.

````python
set(text_2.lower().split(' '))
````

## Exercise 10 (3 points)

Consider the following text:

```
Ethics are built right into the ideals and objectives of the United Nations 
#UNSG @NY Society for Ethical Culture bit.ly/2guVelr
```
Create an object called `text_5` that contains all the words in the above text.

````python
text_5 = "Ethics are built right into the ideals and objectives of the United Nations #UNSG @NY Society for Ethical Culture bit.ly/2guVelr".split(' ')
text_5
````

## Exercise 11 (2 points)

Using Exercise 10, report all the words that start with #

````python
[word for word in text_5 if word.startswith('#')]
````

## Exercise 12 (2 points)

Using Exercise 10, report all the words that start with @
